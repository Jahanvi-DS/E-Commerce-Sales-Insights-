```python
# Import Important Libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time 
all_books = []
```

```python
# 1.loops from 1 to 50 (web scraping)
for page_num in range(1, 51):
    # Dynamic URL that will change the page number every loop
    url = f"http://books.toscrape.com/catalogue/page-{page_num}.html"

    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    books = soup.find_all('article', class_='product_pod')

    # Extracting data from all the books on the page
    for book in books:
        title = book.h3.a['title']
        price = book.find('p', class_='price_color').text
        rating = book.find('p', class_='star-rating')['class'][1]

        all_books.append([title, price, rating])

    print(f"Page {page_num} successfully scraped!")
    time.sleep(0.1)
```

```python
# 2. Converting data into dataframe
df_large = pd.DataFrame(all_books, columns=['Product_Name', 'Price', 'Rating'])

print("\n--- Scraping Completed! ---")
print("Total Rows and Columns:", df_large.shape)
```

```python
# 1. Removing currency symbol from df_large price column and converting into float
df_large['Price_Cleaned'] = df_large['Price'].str[1:].astype(float)
```

```python
# 2. converting rating from text to number
rating_map = {
    'One': 1,
    'Two': 2,
    'Three': 3,
    'Four': 4,
    'Five': 5
}
```

```python
# 3. Converting df_large rating to numeric using map function
df_large['Rating_Numeric'] = df_large['Rating'].map(rating_map)
```

```python
# 4. creating a clean dataset by dropping old, dirty column
df_clean_ecommerce = df_large.drop(columns=['Price', 'Rating'])

print("--- 1,000 Rows Cleaned Data Ka View ---")
print(df_clean_ecommerce.head())
```

```python
# save the clean data to a new file
df_clean_ecommerce.to_csv("cleaned_ecommerce_1000.csv", index=False)
```

```python
# creating order data
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
# 1. give unique identity
df_clean_ecommerce['Product_ID'] = [f"PROD{1000 + i}" for i in range(len(df_clean_ecommerce))]
```

```python
# 2. Setup (for number of order)
num_orders = 1000
order_list = []

# list of product id
all_product_ids = df_clean_ecommerce['Product_ID'].values
```

```python
# 3. Simple Loop (row creation)
for i in range(num_orders):
    order_id = f"ORD{50000 + i}"       # ORD50000, ORD50001, ORD50002...
    customer_id = f"CUST{100 + (i % 50)}"
    prod_id = np.random.choice(all_product_ids)
    quantity = np.random.randint(1, 4)

    # put all order in list
    order_list.append([order_id, customer_id, prod_id, quantity])
```

```python
# 4. DataFrame creation
df_orders = pd.DataFrame(order_list, columns=['OrderID', 'CustomerID', 'Product_ID', 'Quantity'])

print("--- order table is ready! ---")
print(df_orders.head())
```
